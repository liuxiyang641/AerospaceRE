```python
import json
import os
import random
import shutil
```

```python
inverse_relation = {
    'target_of': 'have_target',
    'NA': 'NA',
    'domain_of': 'have_domain',
    'performed_by': 'perform',
    'location_of': 'located_in',
    'applied_in': 'apply',
    'part_of': 'have_part',
    'aspect_of': 'have_aspect',
    'affected_by': 'affect',
    'cooperate_with': 'cooperate_with',
    'type_of': 'include_type',
    'occur_in': 'have_event',
    'result_of': 'have_result',
    'owned_by': 'own',
    'compete_with': 'compete_with',
}
```

```python
k_shot = 8
k_shot_id = 1
```

```python
with open('train.json', 'r') as f:
    all_lines = f.readlines()
    train_data = [json.loads(line) for line in all_lines]
    train_data_by_rel = {}
    for data in train_data:
        rel = data['relation']
        if rel not in train_data_by_rel:
            train_data_by_rel[rel] = []
        train_data_by_rel[rel].append(data)

with open('val.json', 'r') as f:
    all_lines = f.readlines()
    valid_data = [json.loads(line) for line in all_lines]
    valid_data_by_rel = {}
    for data in valid_data:
        rel = data['relation']
        if rel not in valid_data_by_rel:
            valid_data_by_rel[rel] = []
        valid_data_by_rel[rel].append(data)


random.seed(42)
train_samples = []
valid_samples = []

for rel in train_data_by_rel:
    train_samples.extend(random.sample(train_data_by_rel[rel], min(k_shot, len(train_data_by_rel[rel]))))
    valid_samples.extend(random.sample(valid_data_by_rel[rel], min(k_shot, len(valid_data_by_rel[rel]))))


if not os.path.exists(f'k-shot/{k_shot}-{k_shot_id}'):
    os.makedirs(f'k-shot/{k_shot}-{k_shot_id}')
with open(f'k-shot/{k_shot}-{k_shot_id}/train.json', 'w') as f:
    f.write('\n'.join(json.dumps(sample, ensure_ascii=False) for sample in train_samples))

with open(f'k-shot/{k_shot}-{k_shot_id}/val.json', 'w') as f:
    f.write('\n'.join(json.dumps(sample, ensure_ascii=False) for sample in valid_samples))


shutil.copy('test.json', f'k-shot/{k_shot}-{k_shot_id}/test.json')
```
